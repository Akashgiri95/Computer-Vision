```python
# Visualize the Predicted image function
import cv2
import matplotlib.pyplot as plt

def show_image(image_path):
    """
    Function to display an image using matplotlib.
    Works in any Python environment (not just Google Colab).

    Parameters:
    - image_path (str): Path to the image file.
    """
    image = cv2.imread(image_path)
    if image is None:
        print("Error: Image not found!")
        return
    
    # Convert BGR (OpenCV format) to RGB (matplotlib format)
    image_rgb = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)
    
    # Display the image
    plt.figure(figsize=(10, 8))
    plt.imshow(image_rgb)
    plt.axis('off')  # Hide axes
    plt.show()
```

# Nano Model

```python
from ultralytics import YOLO

# Load the YOLO model
model = YOLO('yolo11n.pt')  
image_path = r"/Users/akashgiri/Documents/AIDTM/Trim 3/Computer Vision/External Source/4.Object Detection/roadcross.jpg"
results = model.predict(source=image_path, save=True, conf=0.25)

# Display the results 
import matplotlib.pyplot as plt
import matplotlib.image as mpimg

# Display the original image with predictions
img = mpimg.imread(results[0].save_dir + '/roadcross.jpg')  # Adjust path if needed
plt.figure(figsize=(12, 8))
plt.imshow(img)
plt.axis('off')
plt.show()
```
